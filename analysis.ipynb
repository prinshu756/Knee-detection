```python
import numpy as np 
import pandas as pd
import matplotlib.pyplot as plt
import torch 

from PIL import Image
from pathlib import Path
```

```python
def npy_to_image(npy_path, output_path=None, cmap="gray"):
    """Convert a .npy array to an image file."""
    arr = np.load(npy_path)
    
    # Remove channel dim if present: (1, H, W) -> (H, W)
    if arr.ndim == 3 and arr.shape[0] == 1:
        arr = arr[0]
    
    # Normalize to [0, 255] uint8
    arr = arr.astype(np.float32)
    arr = (arr - arr.min()) / (arr.max() - arr.min() + 1e-8)
    arr = (arr * 255).astype(np.uint8)
    
    img = Image.fromarray(arr, mode="L")  # "L" = grayscale
    
    if output_path:
        img.save(output_path)
        print(f"Saved: {output_path}")
    return img

# Example: convert one file
npy_to_image("train/axial/0000.npy", "train/train_image/0000.png")

# Example: batch convert a folder
for npy_file in Path("train/axial").glob("*.npy"):
    out = npy_file.with_suffix(".png")
    npy_to_image(npy_file, out)
```
