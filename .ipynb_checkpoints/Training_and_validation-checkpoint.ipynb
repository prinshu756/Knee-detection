# Training & Validation

HPO (Grid / Randomized / Bayesian-Optuna) -> full-epoch training -> GPU joblib pipeline + API.

```python
import os
import io
import copy
import glob
import time
import tempfile

import joblib
import numpy as np
import pandas as pd
import torch
import torch.nn.functional as Fn
import matplotlib.pyplot as plt
from torch import nn
from torch.utils.data import Dataset, DataLoader
import torchvision.models as tvm
import torchvision.transforms.functional as TF
from torchvision.transforms import InterpolationMode
from sklearn.model_selection import ParameterGrid, ParameterSampler
import optuna

SEED = 42
IMG_DIR = 'images'
PLANES = ('axial', 'coronal', 'sagittal')
TASKS = ('abnormal', 'acl', 'meniscus')
BATCH_SIZE = 1
LR = 1e-3
EPOCHS = 50
PATIENCE = 5
BACKBONE = 'resnet18'
OUT_DIR = 'outputs'
MODE = os.environ.get('MRNET_MODE', 'demo')   # demo -> fast verify | single -> 1 combo full | all -> 9 combos
HPO_BUDGET = {'demo': {'grid': 4, 'random': 4, 'optuna': 6, 'epochs': 2, 'train_n': 24, 'val_n': 12},
              'single': {'grid': 8, 'random': 8, 'optuna': 12, 'epochs': 2, 'train_n': 96, 'val_n': 48},
              'all': {'grid': 12, 'random': 10, 'optuna': 15, 'epochs': 3, 'train_n': 192, 'val_n': 96}}[MODE]
DEVICE = 'cuda' if torch.cuda.is_available() else 'cpu'
AMP = DEVICE == 'cuda'

torch.backends.cudnn.benchmark = True
os.makedirs(OUT_DIR, exist_ok=True)
np.random.seed(SEED)
torch.manual_seed(SEED)
torch.cuda.manual_seed_all(SEED)
optuna.logging.set_verbosity(optuna.logging.WARNING)
print(f'device: {DEVICE} | AMP: {AMP} | mode: {MODE} | hpo budget: {HPO_BUDGET}')
```

Preprocessing, augmentation, dataset (shared with the serving pipeline).

```python
def standardize(arr):
    arr = arr.astype(np.float32)
    return (arr - arr.mean()) / (arr.std() + 1e-8)


def preprocess(arr, target=32):
    arr = standardize(arr)
    t = torch.from_numpy(np.ascontiguousarray(arr))[None, None]
    arr = Fn.interpolate(t, size=(target, arr.shape[1], arr.shape[2]),
                         mode='trilinear', align_corners=False)[0, 0].numpy()
    return torch.from_numpy(np.stack([arr] * 3))


def load_volume(path_or_bytes, target=32):
    if isinstance(path_or_bytes, (bytes, bytearray)):
        arr = np.load(io.BytesIO(path_or_bytes))
    else:
        arr = np.load(path_or_bytes)
    return preprocess(arr, target)
```

```python
def _slice_batch(vol):
    d, h, w = vol.shape[1:]
    return vol.permute(1, 0, 2, 3).reshape(-1, 1, h, w)


def _unslice_batch(imgs, c, d):
    h, w = imgs.shape[-2:]
    return imgs.reshape(d, c, h, w).permute(1, 0, 2, 3)


class RandomHFlip:
    def __init__(self, p=0.5):
        self.p = p

    def __call__(self, vol):
        return torch.flip(vol, dims=[-1]) if torch.rand(1).item() < self.p else vol


class RandomRotation:
    def __init__(self, degrees=15.0, p=0.5):
        self.degrees, self.p = degrees, p

    def __call__(self, vol):
        if torch.rand(1).item() >= self.p or self.degrees == 0:
            return vol
        ang = (torch.rand(1).item() * 2 - 1) * self.degrees
        imgs = TF.rotate(_slice_batch(vol), ang, interpolation=InterpolationMode.BILINEAR, fill=0.0)
        return _unslice_batch(imgs, vol.shape[0], vol.shape[1])


class RandomTranslation:
    def __init__(self, fraction=0.1, p=0.5):
        self.fraction, self.p = fraction, p

    def __call__(self, vol):
        if torch.rand(1).item() >= self.p or self.fraction == 0:
            return vol
        dx = int(round((torch.rand(1).item() * 2 - 1) * self.fraction * vol.shape[-1]))
        dy = int(round((torch.rand(1).item() * 2 - 1) * self.fraction * vol.shape[-2]))
        imgs = TF.affine(_slice_batch(vol), angle=0, translate=(dx, dy), scale=1.0,
                         shear=[0.0, 0.0], interpolation=InterpolationMode.BILINEAR, fill=0.0)
        return _unslice_batch(imgs, vol.shape[0], vol.shape[1])


def make_transforms(degrees=15.0, translation=0.1):
    return RandomHFlip(0.5), RandomRotation(degrees, 0.5), RandomTranslation(translation, 0.5)


class MRNetDataset(Dataset):
    def __init__(self, root=IMG_DIR, task='abnormal', split='train', plane='sagittal',
                 degrees=15.0, translation=0.1, augment=False):
        self.folder = os.path.join(root, split, plane)
        df = pd.read_csv(os.path.join(root, f'{split}-{task}.csv'), header=None, names=['id', 'label'])
        df['id'] = df['id'].astype(int).astype(str).str.zfill(4)
        self.ids = df['id'].tolist()
        self.labels = df['label'].astype(np.float32).tolist()
        hflip, rot, trans = make_transforms(degrees, translation)
        self.transform = (lambda v: trans(rot(hflip(v)))) if augment else None
        pos, neg = sum(self.labels), len(self.labels) - sum(self.labels)
        self.weights = torch.tensor([1.0, neg / max(pos, 1.0)])

    def __len__(self):
        return len(self.ids)

    def __getitem__(self, idx):
        vol = preprocess(np.load(os.path.join(self.folder, self.ids[idx] + '.npy')))
        if self.transform is not None:
            vol = self.transform(vol)
        return vol, torch.tensor(self.labels[idx])
```

Model + portable joblib payload (state dict + config); loading rebuilds a GPU-warmed fp16 pipeline.

```python
class MRNet2D(nn.Module):
    def __init__(self, backbone='resnet18', pretrained=True):
        super().__init__()
        if backbone == 'resnet18':
            m = tvm.resnet18(weights=tvm.ResNet18_Weights.IMAGENET1K_V1 if pretrained else None)
            m.fc = nn.Identity()
            self.backbone, self.feat_dim = nn.Sequential(*list(m.children())), 512
        elif backbone == 'alexnet':
            m = tvm.alexnet(weights=tvm.AlexNet_Weights.IMAGENET1K_V1 if pretrained else None)
            self.backbone, self.feat_dim = nn.Sequential(m.features, nn.AdaptiveAvgPool2d(1), nn.Flatten()), 256
        else:
            raise ValueError(backbone)
        self.fc = nn.Linear(2 * self.feat_dim, 1)

    def forward(self, x):                     # (B, 3, D, H, W)
        b, c, d, h, w = x.shape
        f = self.backbone(x.permute(0, 2, 1, 3, 4).reshape(b * d, c, h, w)).view(b, d, self.feat_dim)
        return self.fc(torch.cat([f.max(1).values, f.logsumexp(1)], dim=1))


class MRNetPipeline:
    def __init__(self, task, plane, model, threshold=0.5, device=None, amp=True):
        self.task, self.plane, self.threshold = task, plane, threshold
        self.device = device or ('cuda' if torch.cuda.is_available() else 'cpu')
        self.amp = self.device == 'cuda' and amp
        self.model = model.to(self.device).eval()

    def warm(self, device=None):
        if device:
            self.device = device
            self.model = self.model.to(device).eval()
        self.amp = self.device == 'cuda' and self.amp
        x = torch.randn(1, 3, 32, 256, 256, device=self.device)
        with torch.inference_mode():
            self.model(x)
        return self

    @torch.inference_mode()
    def predict_proba(self, vol):
        if not torch.is_tensor(vol):
            vol = load_volume(vol)
        x = vol.unsqueeze(0).to(self.device)
        with torch.amp.autocast('cuda', enabled=self.amp):
            return float(torch.sigmoid(self.model(x)).item())

    def predict(self, vol):
        p = self.predict_proba(vol)
        pos = p >= self.threshold
        return {'task': self.task, 'plane': self.plane, 'probability': round(p, 4),
                'label': int(pos), 'class': self.task if pos else 'normal'}


def save_pipeline(model, task, plane, out_dir=OUT_DIR, name=None):
    path = os.path.join(out_dir, name or f'model_{task}_{plane}.joblib')
    payload = {'backbone': 'resnet18', 'task': task, 'plane': plane, 'threshold': 0.5,
               'model_state': {k: v.detach().cpu() for k, v in model.state_dict().items()}}
    joblib.dump(payload, path, compress=3)
    return path


def load_pipeline(path):
    payload = joblib.load(path)
    model = MRNet2D(payload['backbone'], pretrained=False)
    model.load_state_dict(payload['model_state'])
    return MRNetPipeline(payload['task'], payload['plane'], model, threshold=payload['threshold']).warm()
```

```python
model = MRNet2D(BACKBONE).to(DEVICE)
print('params_M:', round(sum(p.numel() for p in model.parameters()) / 1e6, 2), '| on:', next(model.parameters()).device)
probe = MRNetDataset(task='acl', split='train', plane='sagittal', augment=True)
print('train pos_weight:', probe.weights.tolist(), f'| positives: {sum(probe.labels):.0f}/{len(probe)}')
```

Loop: epochs -> one-exam batches -> weighted BCE -> AMP backward -> validation + early stopping.

```python
def auc_score(labels, scores):
    y = np.asarray(labels, dtype=np.float64)
    r = pd.Series(np.asarray(scores, dtype=np.float64)).rank(method='average').values
    pos, neg = y.sum(), len(y) - y.sum()
    if pos == 0 or neg == 0:
        return float('nan')
    return float((r[y == 1].sum() - pos * (pos + 1) / 2) / (pos * neg))


def train_one_epoch(model, loader, criterion, optimizer, scaler):
    model.train()
    total, n = 0.0, 0
    for x, y in loader:                      # x: (1, 3, D, H, W) = one exam
        x, y = x.to(DEVICE), y.float().to(DEVICE)
        optimizer.zero_grad(set_to_none=True)
        with torch.amp.autocast('cuda', enabled=AMP):
            logits = model(x).view(-1)
            loss = criterion(logits, y)
        scaler.scale(loss).backward()
        scaler.step(optimizer)
        scaler.update()
        total += loss.item() * len(y)
        n += len(y)
    return total / max(n, 1)


@torch.inference_mode()
def evaluate(model, loader, criterion):
    model.eval()
    total, n, ys, ps = 0.0, 0, [], []
    for x, y in loader:
        x, y = x.to(DEVICE), y.float().to(DEVICE)
        with torch.amp.autocast('cuda', enabled=AMP):
            logits = model(x).view(-1)
        total += criterion(logits.float(), y).item() * len(y)
        n += len(y)
        ys.append(y.cpu())
        ps.append(torch.sigmoid(logits.float()).cpu())
    y, p = torch.cat(ys), torch.cat(ps)
    return total / max(n, 1), auc_score(y, p), float((((p >= 0.5).float() == y).float().mean()))
```

HPO: exhaustive grid (`ParameterGrid`), randomized (`ParameterSampler`), Bayesian TPE (Optuna) - same short-budget objective, cached volumes.

```python
PARAM_SPACE = {'lr': [1e-4, 3e-4, 1e-3], 'weight_decay': [0.0, 1e-4], 'freeze_backbone': [False, True],
               'degrees': [0.0, 15.0, 25.0], 'translation': [0.0, 0.1]}
RANDOM_SPACE = {'lr': (1e-5, 3e-3), 'weight_decay': (1e-6, 1e-2), 'degrees': (0.0, 25.0)}


def _vol_cache(task, plane, split, ids=None):
    ds = MRNetDataset(task=task, split=split, plane=plane)
    keep = ids if ids is not None else ds.ids
    idx = {i: k for k, i in enumerate(ds.ids)}
    out = []
    for eid in keep:
        arr = np.load(os.path.join(IMG_DIR, split, plane, eid + '.npy'))
        out.append((preprocess(arr), torch.tensor(ds.labels[idx[eid]])))
    return out


class TrialRunner:
    def __init__(self, task, plane, train_n, val_n, epochs):
        rng = np.random.default_rng(SEED)
        tr = MRNetDataset(task=task, split='train', plane=plane)
        va = MRNetDataset(task=task, split='valid', plane=plane)
        self.train_ids = sorted(rng.choice(tr.ids, size=min(train_n, len(tr.ids)), replace=False).tolist())
        self.val_ids = sorted(rng.choice(va.ids, size=min(val_n, len(va.ids)), replace=False).tolist())
        self.train = _vol_cache(task, plane, 'train', self.train_ids)
        self.val = _vol_cache(task, plane, 'valid', self.val_ids)
        self.epochs = epochs
        pw = tr.weights[1].to(DEVICE)
        self.criterion = {'pw': pw}

    def run(self, params, epochs=None, verbose=False):
        epochs = epochs or self.epochs
        pw = self.criterion['pw']
        hflip, rot, trans = make_transforms(float(params['degrees']), float(params['translation']))

        def apply_t(vol):                     # per-exam augmentation, seeded order via shuffle below
            return trans(rot(hflip(vol)))

        model = MRNet2D(BACKBONE).to(DEVICE)
        if params.get('freeze_backbone'):
            for p in model.backbone.parameters():
                p.requires_grad = False
        criterion = nn.BCEWithLogitsLoss(pos_weight=pw)
        optimizer = torch.optim.Adam(filter(lambda p: p.requires_grad, model.parameters()),
                                     lr=float(params['lr']), weight_decay=float(params['weight_decay']))
        scaler = torch.amp.GradScaler('cuda', enabled=AMP)
        order = list(range(len(self.train)))
        rng = np.random.default_rng(SEED)
        best_auc = float('nan')
        for _ in range(epochs):
            model.train()
            rng.shuffle(order)
            for i in order:
                x = apply_t(self.train[i][0]).unsqueeze(0).to(DEVICE)
                y = self.train[i][1].float().view(1).to(DEVICE)
                optimizer.zero_grad(set_to_none=True)
                with torch.amp.autocast('cuda', enabled=AMP):
                    loss = criterion(model(x).view(-1), y)
                scaler.scale(loss).backward()
                scaler.step(optimizer)
                scaler.update()
        vl, va, _ = evaluate_pairs(model, self.val, criterion)
        if verbose:
            print(f'  trial {params} -> val_loss {vl:.3f} auc {va:.3f}')
        return va


@torch.inference_mode()
def evaluate_pairs(model, pairs, criterion):
    model.eval()
    ys, ps, tot = [], [], 0.0
    for vol, y in pairs:
        x = vol.unsqueeze(0).to(DEVICE)
        with torch.amp.autocast('cuda', enabled=AMP):
            logits = model(x).view(-1)
        tot += criterion(logits.float(), y.float().view(-1).to(DEVICE)).item()
        ps.append(torch.sigmoid(logits.float()).cpu())
        ys.append(y)
    y, p = torch.stack(ys).float(), torch.cat(ps)
    return tot / len(pairs), auc_score(y, p), float((((p >= 0.5).float() == y).float().mean()))


def run_hpo(task, plane, budget):
    runner = TrialRunner(task, plane, budget['train_n'], budget['val_n'], budget['epochs'])
    rows, t0 = [], time.time()

    grid = list(ParameterGrid({k: v for k, v in PARAM_SPACE.items()
                               if k in ('lr', 'weight_decay', 'degrees')}))[:budget['grid']]
    for p in grid:
        p = dict(p, freeze_backbone=False, translation=0.1)
        rows.append({'method': 'grid', **p, 'val_auc': runner.run(p)})

    sampler = list(ParameterSampler(RANDOM_SPACE, n_iter=budget['random'], random_state=SEED))
    rng_fb = np.random.default_rng(SEED)
    for p in sampler:
        p = dict(lr=float(p['lr']), weight_decay=float(p['weight_decay']), degrees=float(p['degrees']),
                 freeze_backbone=bool(rng_fb.integers(2)), translation=0.1)
        rows.append({'method': 'random', **p, 'val_auc': runner.run(p)})

    def objective(trial):
        p = {'lr': trial.suggest_float('lr', *RANDOM_SPACE['lr'], log=True),
             'weight_decay': trial.suggest_float('weight_decay', *RANDOM_SPACE['weight_decay'], log=True),
             'degrees': trial.suggest_float('degrees', *RANDOM_SPACE['degrees']),
             'freeze_backbone': trial.suggest_categorical('freeze_backbone', [False, True]),
             'translation': trial.suggest_categorical('translation', [0.0, 0.1])}
        return runner.run(p)

    study = optuna.create_study(direction='maximize', sampler=optuna.samplers.TPESampler(seed=SEED))
    study.optimize(objective, n_trials=budget['optuna'], show_progress_bar=False)
    for p in study.trials:
        rows.append({'method': 'bayes', **p.params, 'val_auc': p.value})

    df = pd.DataFrame(rows).sort_values('val_auc', ascending=False).reset_index(drop=True)
    print(df.to_string(index=False))
    print(f'HPO done: {len(df)} trials in {time.time() - t0:.0f}s')
    best = df.iloc[0].to_dict()
    best = {k: (int(v) if k == 'freeze_backbone' else float(v)) for k, v in best.items() if k not in ('method', 'val_auc')}
    print('best config:', best, '| val_auc', round(df.iloc[0]['val_auc'], 4))
    return best, df
```

```python
TASK, PLANE = 'acl', 'sagittal'
best_params, hpo_table = run_hpo(TASK, PLANE, HPO_BUDGET)
hpo_table.to_csv(os.path.join(OUT_DIR, f'hpo_{MODE}.csv'), index=False)
```

```python
def train_model(task, plane, params=None, backbone=BACKBONE, epochs=None, patience=None,
                limit_train=None, limit_val=None, save_name=None, pretrained=True, tag=''):
    params = params or {}
    epochs, patience = epochs or EPOCHS, patience or PATIENCE
    train_ds = MRNetDataset(task=task, split='train', plane=plane, augment=True,
                            degrees=params.get('degrees', 15.0), translation=params.get('translation', 0.1))
    valid_ds = MRNetDataset(task=task, split='valid', plane=plane)
    if limit_train:
        idx = np.random.default_rng(SEED).choice(len(train_ds), size=min(limit_train, len(train_ds)), replace=False)
        train_ds = torch.utils.data.Subset(train_ds, idx.tolist())
    if limit_val:
        idx = np.random.default_rng(SEED).choice(len(valid_ds), size=min(limit_val, len(valid_ds)), replace=False)
        valid_ds = torch.utils.data.Subset(valid_ds, idx.tolist())
    g = torch.Generator().manual_seed(SEED)
    train_loader = DataLoader(train_ds, batch_size=BATCH_SIZE, shuffle=True, num_workers=0, pin_memory=True, generator=g)
    valid_loader = DataLoader(valid_ds, batch_size=BATCH_SIZE, shuffle=False, num_workers=0, pin_memory=True)
    pos_weight = MRNetDataset(task=task, split='train', plane=plane).weights[1].to(DEVICE)
    print(f'[{task}/{plane}] train={len(train_ds)} valid={len(valid_ds)} pos_weight={pos_weight.item():.2f} | params={params}')

    model = MRNet2D(backbone, pretrained).to(DEVICE)
    if params.get('freeze_backbone'):
        for p in model.backbone.parameters():
            p.requires_grad = False
        print(f'frozen backbone -> trainable_M: '
              f'{sum(p.numel() for p in model.parameters() if p.requires_grad) / 1e6:.2f}')
    criterion = nn.BCEWithLogitsLoss(pos_weight=pos_weight)
    optimizer = torch.optim.Adam(filter(lambda p: p.requires_grad, model.parameters()),
                                 lr=float(params.get('lr', LR)), weight_decay=float(params.get('weight_decay', 0.0)))
    scaler = torch.amp.GradScaler('cuda', enabled=AMP)
    history = {'train_loss': [], 'val_loss': [], 'val_auc': [], 'val_acc': []}
    best_val, best_state, wait = float('inf'), None, 0

    for epoch in range(1, epochs + 1):
        t0 = time.time()
        tl = train_one_epoch(model, train_loader, criterion, optimizer, scaler)
        vl, va, vac = evaluate(model, valid_loader, criterion)
        for k, v in zip(history, (tl, vl, va, vac)):
            history[k].append(v)
        flag = ''
        if vl < best_val:
            best_val, best_state, wait, flag = vl, copy.deepcopy(model.state_dict()), 0, ' *best*'
        else:
            wait += 1
        print(f'epoch {epoch:3d}/{epochs} | train {tl:.4f} | val {vl:.4f} | auc {va:.3f} '
              f'| acc {vac:.3f} | {time.time() - t0:.1f}s{flag}', flush=True)
        if wait >= patience:
            print(f'early stopping at epoch {epoch} (no improvement for {patience} epochs)')
            break

    model.load_state_dict(best_state)
    save_name = save_name or f'best_{task}_{plane}{tag}.pth'
    pth_path = os.path.join(OUT_DIR, save_name)
    torch.save({'task': task, 'plane': plane, 'params': params, 'epoch': len(history['val_loss']),
                'val_loss': best_val, 'model_state': best_state}, pth_path)
    jl_path = save_pipeline(model, task, plane,
                            name=os.path.basename(save_name).replace('best_', 'model_').rsplit('.', 1)[0] + '.joblib')
    print(f'saved {pth_path} + {jl_path} | best val_loss {best_val:.4f}')
    return model, history, {'pth': pth_path, 'joblib': jl_path}
```

```python
RUN_FINAL = True
TAG = '' if MODE != 'demo' else '_demo'
if RUN_FINAL:
    if MODE == 'demo':
        model, history, artifacts = train_model(TASK, PLANE, params=best_params, epochs=EPOCHS,
                                                patience=EPOCHS + 1, limit_train=24, limit_val=12,
                                                save_name=f'best_{TASK}_{PLANE}_demo.pth')
    else:
        model, history, artifacts = train_model(TASK, PLANE, params=best_params, epochs=EPOCHS,
                                                patience=PATIENCE)
    results = pd.DataFrame([{'task_plane': f'{TASK}/{PLANE}', 'mode': MODE, 'epochs_run': len(history['val_loss']),
                             'best_val_loss': round(min(history['val_loss']), 4),
                             'best_val_auc': round(max(history['val_auc']), 4),
                             'lr': best_params['lr'], 'weight_decay': best_params['weight_decay'],
                             'freeze_backbone': best_params['freeze_backbone'],
                             'degrees': best_params['degrees'], 'translation': best_params['translation'],
                             'joblib': os.path.basename(artifacts['joblib'])}])
    results.to_csv(os.path.join(OUT_DIR, f'training_metrics{TAG or "_" + MODE}.csv'), index=False)
    print(results.to_string(index=False))

    fig, ax = plt.subplots(figsize=(7, 4))
    ax.plot(range(1, len(history['val_loss']) + 1), history['val_loss'], marker='o', label='val')
    ax.plot(range(1, len(history['train_loss']) + 1), history['train_loss'], marker='.', label='train')
    ax.set_xlabel('epoch')
    ax.set_ylabel('weighted BCE')
    ax.legend()
    ax.set_title('early stopping on validation loss')
    fig.savefig(os.path.join(OUT_DIR, f'loss_curves{TAG or "_" + MODE}.png'), dpi=120, bbox_inches='tight')
    plt.show()
```

```python
from fastapi import FastAPI, File, HTTPException, UploadFile

app = FastAPI(title='MRNet Knee MRI API', version='1.2')
_pipes = {}


def find_model(task, plane):
    p = os.path.join(OUT_DIR, f'model_{task}_{plane}.joblib')
    if not os.path.isfile(p):
        hits = sorted(glob.glob(os.path.join(OUT_DIR, f'model_{task}_{plane}*.joblib')), key=os.path.getmtime)
        p = hits[-1] if hits else None
    return p


def get_pipe(task, plane):
    key = f'{task}/{plane}'
    if key not in _pipes:
        if (p := find_model(task, plane)) is None:
            raise HTTPException(404, f"no trained model for task='{task}' plane='{plane}'")
        _pipes[key] = load_pipeline(p)
    return _pipes[key]


@app.get('/health')
def health():
    return {'status': 'ok', 'models_available': [f'{t}/{p}' for t in TASKS for p in PLANES if find_model(t, p)]}


@app.post('/predict')
def predict(task: str, plane: str, file: UploadFile = File(...)):
    if task not in TASKS or plane not in PLANES:
        raise HTTPException(400, f'task must be one of {TASKS}, plane one of {PLANES}')
    try:
        return get_pipe(task, plane).predict(file.file.read())
    except (OSError, ValueError) as exc:
        raise HTTPException(400, f'bad .npy upload: {exc}')

# live:  uvicorn Training_and_validation:app --port 8000   (or: import uvicorn; uvicorn.run(app, port=8000))
```

```python
from fastapi.testclient import TestClient

client = TestClient(app)
r = client.get('/health')
print('/health', r.status_code, r.json())
assert r.status_code == 200 and f'{TASK}/{PLANE}' in r.json()['models_available']
with open(os.path.join(IMG_DIR, 'valid', 'sagittal', '1130.npy'), 'rb') as fh:
    r = client.post('/predict', params={'task': TASK, 'plane': PLANE},
                    files={'file': ('1130.npy', fh, 'application/octet-stream')})
print('/predict', r.status_code, r.json())
assert r.status_code == 200 and 0.0 <= r.json()['probability'] <= 1.0
r404 = client.post('/predict', params={'task': TASK, 'plane': 'axial'},
                   files={'file': ('x.npy', npy_bytes, 'application/octet-stream')})
assert r404.status_code == 404
r400 = client.post('/predict', params={'task': TASK, 'plane': PLANE},
                   files={'file': ('bad.npy', b'garbage', 'application/octet-stream')})
assert r400.status_code == 400
print('API verified: 200 / 404 / 400 + GPU-warmed pipeline reuse')
```

```python
try:
    import ipywidgets as widgets
    from IPython.display import display

    def _on_upload(change):
        for meta in change['new']:
            t0 = time.time()
            print(f"{meta['name']} -> {pipe.predict(meta['content'])} ({1000 * (time.time() - t0):.0f} ms, GPU)")

    up = widgets.FileUpload(accept='.npy', multiple=True)
    up.observe(_on_upload, names='value')
    dd = widgets.Dropdown(options=[f'{t}/{p}' for t in TASKS for p in PLANES],
                          value=f'{TASK}/{PLANE}', description='model:')

    def _on_model(change):
        global pipe
        pipe = get_pipe(*change['new'].split('/'))

    dd.observe(_on_model, names='value')
    display(widgets.VBox([dd, up]))
except ImportError:
    print('ipywidgets not available - use pipe.predict(bytes) or the /predict API')

# full sweep: MRNET_MODE=all python -m jupyter nbconvert --to notebook --execute --inplace Training_and_validation.ipynb
```
